```python
# Imporitng pandas as pd
import pandas as pd
```

```python
# Load data from an csv file and create a dataframe
df = pd.read_csv('data (1).csv')
print(df.to_string())
```

```python
# print 5 from head
print(df.head())
```

```python
# Print last 5 row column
print(df.tail())
```

```python
#total row column count
print(df.shape)
```

```python
# all information show
print(df.info())
```

```python
# Check duplicate row
print(df.duplicated())
```

```python
# Remove empty sell if any exists(creating new dataframe)
new_df = df.dropna()
print(new_df.to_string())
```

```python
# remove null value from original dataframe
df.dropna(inplace = True)
print(df.to_string())
```

```python
# Calculating mean, median and mode
x = df["Calories"].mean()
y = df["Calories"].median()
z = df["Pulse"].mode()[0]
print("mean: ",x)
print("median: ",y)
print("mode: ",z)
```

```python
# Show correlation between columns
print(df.corr())
```

```python
# Show a graphic representation
import matplotlib.pyplot as plt

df.plot()
plt.show()
```

```python
# histogram
df["Duration"].plot(kind = 'hist')
```

```python
# outlier check
df.boxplot(column=['Pulse'])
plt.show()
```

```python
# data filter and selection
high_pulse = df.loc[df['Pulse'] > 100]
print(high_pulse)

subset = df.iloc[0:3, 0:2]
print(subset)
```

```python
# data cleaning
calories_mean = df['Calories'].mean()
df['Calories'] = df['Calories'].fillna(calories_mean)
df.drop_duplicates(inplace=True)
print(df)
```

```python
# Creating a table or dataframe using(Datframe function)
mydataset = {
  'cars': ["BMW", "Volvo", "Ford"],
  'passings': [3, 7, 2]
}

myvar = pd.DataFrame(mydataset)
print(myvar)
```

```python
# combine dataframe
df1 = pd.DataFrame({'Name': ['A', 'B'], 'Score': [80, 90]})
df2 = pd.DataFrame({'Name': ['C', 'D'], 'Score': [85, 95]})

combined_df = pd.concat([df1, df2], ignore_index=True)
print(combined_df)
```
